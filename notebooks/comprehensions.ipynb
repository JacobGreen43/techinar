# List Comprehensions

A list comprehension is a concise, shorthand syntax in Python used to create a new list out of an existing iterable.

## What is an iterable?

An iterable is anything that can be iterated on (looped over) using a `for` loop. The iterable will provide one value at a time to the loop, but not necessarily in a fixed sequence.

## Examples of iterables

### Lists

A list has a definite order, and it can be modified.

```python
groceries = ["apples", "bread", "chicken", "potatoes"]
groceries.insert(1, "bananas")
for item in groceries:
    print(item)
```

### Tuples

A tuple has a definite order, but it cannot be modified.

```python
coordinates = (23.5, 45.2, 67.8)
for c in coordinates:
    print(c)
```

### Set

A set has no definite order and has no duplicate elements.

```python
colors = {"red", "green", "blue"}
print("Original set")
for color in colors:
    print("  ", color)

print("Set after adding yellow")
colors.add("yellow")
for color in colors:
    print("  ", color)
    
print("Set after attempting to add green")
colors.add("green")
for color in colors:
    print("  ", color)
```

### Dictionary

A dictionary holds a set of key:value pairs. Values in the dictionary can be accessed using the key. Iterating over the dictionary returns
the keys. The `.items()` method returns an iterable of all of the pairs, and the `.values()` method is just the values.

```python
authors = {"Adventures of Huckleberry Finn": "Mark Twain", 
           "Pride and Prejudice": "Jane Austen", 
           "The Great Gatsby": "F. Scott Fitzgerald",
           "1984": "George Orwell", 
           "The Hobbit": "J.R.R. Tolkien", 
           "The Screwtape Letters": "C.S. Lewis", 
           "The Adventures of Sherlock Holmes": "Arthur Conan Doyle"}
b = "The Hobbit"
print(f"{b} was written by {authors[b]}")
print("All books.")
for book in authors:
    print("  ", book)
print("All authors.")
for author in authors.values():
    print("  ", author)
print("All books and their authors.")
for book, author in authors.items():
    print(f"  {book} was written by {author}.")
```

### Strings

Strings are iterable sequences of characters, so they can be used in for loops.

```python
for letter in "Hobbit":
    print(letter)
```

### range() function

The range function produces a sequence that is an iterable. We have seen it used many times in for loops.

### Generator functions

A generator function produces multiple values and can be used in a loop.

```python
names = ["Simón", "Garcia", "Cucurella", "Baena", "Pedri", "Olmo", "Yamal", "Oyarzabel"]
numbers = [23, 4, 24, 15, 20, 10, 19, 21]
for name, number in zip(names, numbers):
    print(f"{name} wears number {number}.")
```

## List Comprehension Form

new_list = **[** *expression* **for** *item* **in** *iterable* **if** *condition* **]**

To understand how Python evaluates this, it helps to basically read it from right to left:

1. *iterable*: The original collection or sequence you are looping through.
2. *item*: The loop variable that grabs each element one at a time.
3. *condition* (optional): A filter. If the item meets this criteria, it is included; otherwise, it is skipped.
4. *expression*: The value that actually gets saved into your new list.

Python creates a new list by looping through the iterable. For each item in the iterable, it includes the item if it meets the condition, and then it evaluates the expression using the item. Each of these expressions becomes an element in the new list.

## Examples

### Transformations

Given a list of numbers, create a list that is the double of each number.

#### For loop method

```python
numbers = [23, 4, 24, 15, 20, 10, 19, 21]
doubled = []
for number in numbers:
    doubled.append(number * 2)
print(doubled)
```

#### List comprehension

```python
numbers = [23, 4, 24, 15, 20, 10, 19, 21]
doubled = [number * 2 for number in numbers]
print(doubled)
```

### Calculations

Given a list of x offsets and y offsets, calculate the distances.

```python
import math

dxs = [5, 3, 7, 2, 8, 1, 4, 6]
dys = [2, 4, 1, 3, 5, 7, 6, 8]

# Print the distances using a for loop
offsets = zip(dxs, dys)
for dx, dy in offsets:
    print(f"dx: {dx}, dy: {dy}, distance: {math.sqrt(dx**2 + dy**2)}")

# Create a list of distances using a list comprehension
distances = [math.sqrt(dx**2 + dy**2) for dx, dy in zip(dxs, dys)]
print(distances)

# Create a list of tuples containing dx, dy, and distance using a list comprehension
distance_tuples = [(dx, dy, math.sqrt(dx**2 + dy**2)) for dx, dy in zip(dxs, dys)]
print(distance_tuples)

# Print the tuples using a for loop
for dx, dy, distance in distance_tuples:
    print(f"dx: {dx}, dy: {dy}, distance: {distance}")
```

### Coin tosses

```python
from random import randint

# Using a function
number_of_tosses = 20
def head_or_tails(num):
    if num == 0:
        return "heads"
    else:
        return "tails"
tosses = [head_or_tails(randint(0, 1)) for _ in range(number_of_tosses)]
print(tosses)

# Using a conditional expression
tosses = ["heads" if randint(0, 1) == 0 else "tails" for _ in range(number_of_tosses)]
print(tosses)
```

### Filtering

Given a list of grades, make a list of the ones that are passing.

```python
grades = [88, 55, 76, 92, 67, 81, 43, 95]
passing_grades = [grade for grade in grades if grade >= 60]
print(passing_grades)

# Use a list of tuples to store the student name and grade
student_grades = [("Jack", 88), ("Bob", 55), ("Charlie", 76), ("David", 92), ("Joe", 67), ("Frank", 81), ("Bill", 43), ("Jeremy", 95)]
passing_students = [student for student, grade in student_grades if grade >= 60]
print(passing_students)
```

### Pandas

Use a list comprehension to generate a new column.

```python
import pandas as pd

df = pd.DataFrame({"Name": ["Jack", "Bob", "Charlie", "David", "Joe", "Frank", "Bill", "Jeremy"],
                   "Score": [88, 55, 76, 92, 67, 81, 43, 95]})
print(df)
df["Status"] = ["Pass" if score >= 60 else "Fail" for score in df["Score"]]
print(df)
```
